```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
from skimage.metrics import peak_signal_noise_ratio, structural_similarity
from pathlib import Path
```

```python
def plot_csv(csv_path, save_path, title, vmin=None, vmax=None):
    """
    Loads a 2D numeric grid from a CSV file and saves it as a colormap image.
 
    Parameters:
    ----------
    csv_path : str
        The file path to the input CSV file containing numeric data.
    save_path : str
        The file path where the generated plot image will be saved.
    title : str
        The title displayed at the top of the plot.
    vmin : float, optional
        The minimum data value that maps to the bottom of the color scale.
        Defaults to None (scaled to data minimum).
    vmax : float, optional
        The maximum data value that maps to the top of the color scale.
        Defaults to None (scaled to data maximum).
    """
    # Load numeric data directly
    data = np.genfromtxt(csv_path, delimiter=',', dtype=None, encoding='utf-8')
    
    # Plot the 2D grid directly with a colormap
    plt.figure()
    plt.imshow(data, cmap='viridis', aspect='equal', vmin=vmin, vmax=vmax)
    plt.title(title)
    plt.colorbar()
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    plt.close()
```

```python
# Create an images folder if one doesn't already exist
Path("images").mkdir(parents=True, exist_ok=True)
```

```python
# Load CSV file with automatically inferred data types
plot_csv('./bin/outputs/ohtcfrp_00_vs_ohtcfrp_11_u.csv', './images/ohtcfrp_00_vs_ohtcfrp_11_u.png', 'u displacement', 0, -0.25)
plot_csv('./bin/outputs/ohtcfrp_00_vs_ohtcfrp_11_v.csv', './images/ohtcfrp_00_vs_ohtcfrp_11_v.png', 'v displacement', -1, -2)
plot_csv('./bin/outputs/ohtcfrp_00_vs_ohtcfrp_11_exx.csv', './images/ohtcfrp_00_vs_ohtcfrp_11_exx.png', 'exx strain', -2e-3, 0)
plot_csv('./bin/outputs/ohtcfrp_00_vs_ohtcfrp_11_exy.csv', './images/ohtcfrp_00_vs_ohtcfrp_11_exy.png', 'exy strain', -2e-3, 2e-3)
plot_csv('./bin/outputs/ohtcfrp_00_vs_ohtcfrp_11_eyy.csv', './images/ohtcfrp_00_vs_ohtcfrp_11_eyy.png', 'eyy strain', 5e-3, 10e-3)
```
